```python
# 1. Mount your Google Drive to the Colab Cloud instance
from google.colab import drive
import os, requests

drive.mount('/content/drive')

# Target directory in Google Drive
drive_target = "/content/drive/My Drive/Noir Vault MP4s"
os.makedirs(drive_target, exist_ok=True)

# 2. Fetch all releases from your GitHub Repo
repo = "hipster-doofus/noir-vault-v2"
releases_url = f"https://api.github.com/repos/{repo}/releases"
releases = requests.get(releases_url).json()

# 3. Stream all MP4 release assets directly into Google Drive
for release in releases:
    assets = release.get("assets", [])
    for asset in assets:
        name = asset.get("name", "")
        download_url = asset.get("browser_download_url", "")

        if name.endswith(".mp4"):
            target_file = os.path.join(drive_target, name)
            if not os.path.exists(target_file):
                print(f"Downloading {name} directly to Google Drive...")
                r = requests.get(download_url, stream=True)
                with open(target_file, 'wb') as f:
                    for chunk in r.iter_content(chunk_size=1024*1024):
                        if chunk:
                            f.write(chunk)
                print(f" Done: {name}")

print("All historical clips synced to Google Drive successfully!")
```
